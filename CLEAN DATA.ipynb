```sql
CREATE OR REPLACE TABLE resturant_csv.cleaned_restaurant_data AS
WITH raw_data AS (
    SELECT DISTINCT * 
    FROM resturant_csv.all_restaurants_combined 
)
SELECT 
    CAST(order_id AS BIGINT) AS order_id,
    CAST(order_date AS DATE) AS order_date,
    CAST(hour AS INT) AS order_hour, 
    TRIM(category) AS category,
    TRIM(item_name) AS item_name,
    CAST(price AS DOUBLE) AS price,
    CAST(quantity AS INT) AS quantity,
    CAST(discount AS DOUBLE) AS discount,
    CAST(total_amount AS DOUBLE) AS total_amount,
    COALESCE(branch, 'Unknown') AS branch, 
    COALESCE(payment_method, 'Not Specified') AS payment_method,
    order_type,
    customer_id,
    
    CASE 
        WHEN rating < 1 OR rating > 5 THEN NULL 
        ELSE rating 
    END AS rating,
    
    CASE 
        WHEN is_weekend NOT IN (0, 1) THEN 0 
        ELSE is_weekend 
    END AS is_weekend,

    current_timestamp() AS last_processed_at

FROM raw_data
WHERE order_id IS NOT NULL; 
```

```sql
CREATE OR REPLACE TABLE resturant_csv.restaurant_dates AS
WITH distinct_dates AS (
    SELECT DISTINCT order_date AS date_column
    FROM resturant_csv.all_restaurants_combined
    WHERE order_date IS NOT NULL
)
SELECT 
  date_column,
  DAY(date_column) AS day_of_month,
  DATE_FORMAT(date_column, 'EEEE') AS day_name, 
  DAYOFWEEK(date_column) AS day_of_week, 
  
  WEEKOFYEAR(date_column) AS week_of_year,
  
  MONTH(date_column) AS month_number,
  DATE_FORMAT(date_column, 'MMMM') AS month_name, 
  
  QUARTER(date_column) AS quarter,
  YEAR(date_column) AS year
  
FROM distinct_dates
ORDER BY date_column;
```
